```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = r"C:\Users\acer\OneDrive\Pictures\Camera Roll\Virat Kohli.jpg"

color = cv2.imread(image_path)

if color is None:
    raise FileNotFoundError("Image not found. Check the image path.")

rgb = cv2.cvtColor(color, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(rgb)
plt.title("Original Virat Kohli Color Image")
plt.axis("off")
plt.show()
```

```python
b, g, r = cv2.split(color)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(rgb)
plt.title("Original Color Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.hist(b.ravel(), bins=256, range=(0, 256))
plt.title("Blue Channel Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 3)
plt.hist(g.ravel(), bins=256, range=(0, 256))
plt.title("Green Channel Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.hist(r.ravel(), bins=256, range=(0, 256))
plt.title("Red Channel Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
hsv = cv2.cvtColor(color, cv2.COLOR_BGR2HSV)

h, s, v = cv2.split(hsv)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(h, cmap="gray")
plt.title("Hue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(s, cmap="gray")
plt.title("Saturation Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(v, cmap="gray")
plt.title("Value Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
v_equalized = cv2.equalizeHist(v)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(v, cmap="gray")
plt.title("Original V Channel")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(v_equalized, cmap="gray")
plt.title("Equalized V Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
hsv_enhanced = cv2.merge([h, s, v_equalized])

enhanced_bgr = cv2.cvtColor(
    hsv_enhanced,
    cv2.COLOR_HSV2BGR
)

enhanced_rgb = cv2.cvtColor(
    enhanced_bgr,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(6, 6))
plt.imshow(enhanced_rgb)
plt.title("Virat Kohli - HSV Enhanced Image")
plt.axis("off")
plt.show()
```
